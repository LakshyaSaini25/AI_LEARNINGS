## **NLP Tutorial - Text Representation: TF-IDF**

**What is TF-IDF?**

* TF stands for Term Frequency and denotes the ratio of number of times a particular word appeared in a Document to total number of words in the document.

  **Term Frequency(TF) = [number of times word appeared / total no of words in a document]**

Term Frequency values ranges between 0 and 1. If a word occurs more number of times, then it's value will be close to 1.

* IDF stands for Inverse Document Frequency and denotes the log of ratio of total number of documents/datapoints in the whole dataset to the number of documents that contains the particular word.

   **Inverse Document Frequency(IDF) = [log(Total number of documents / number of documents that contains the word)]**

In IDF, if a word occured in more number of documents and is common across all documents, then it's value will be less and ratio will approaches to 0.

Finally:

   **TF-IDF = Term Frequency(TF) * Inverse Document Frequency(IDF)**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "Thor eating pizza, Loki is eating pizza, Ironman ate pizza already",
    "Apple is announcing new iphone tomorrow",
    "Tesla is announcing new model-3 tomorrow",
    "Google is announcing new pixel-6 tomorrow",
    "Microsoft is announcing new surface tomorrow",
    "Amazon is announcing new eco-dot tomorrow",
    "I am eating biryani and you are eating grapes"
]
```

```python
#let's create the vectorizer and fit the corpus and transform them accordingly
v= TfidfVectorizer()

transformed_output = v.fit_transform(corpus) # Now it will first build the vocabulary and then transform it to out vector representation

print(v.vocabulary_)
```

```python
# Now let's print the idf score for each of the word present in our vocabulary (dictionary)
all_features_names = v.get_feature_names_out()
print(all_features_names)


for word in all_features_names:
    indx = v.vocabulary_.get(word) # get the index of the word from the vocabulary
    print(f"{word} : {v.idf_[indx]}") # getting the idf score using the v.idf_[indx]
```

**We can see that the word like "is" have low idf score as it is appearing in almost all the docs of the corpus and the words like "thor" "iphone" have higher idf score as they are less common among all the docs**

```python
# Checking out Transformed output that contains Vector representation of each doc of the corpus
print(corpus[:2])
print(transformed_output.toarray()[:2])
```

## **E-Commerce Item Catorogy Classification using Tf-Idf**

```python
import pandas as pd

df = pd.read_csv('Ecommerce_data.csv')
print(df.shape)
df.head()
```

```python
# Printing the distinct categories of label we have in our dataframe
print(df.label.value_counts())
```

**We have 4 types of categories of label and all are equally Balanced**

```python
# Great now we convert the labels into Numbers as ML model needs numbers
df['label_num'] = df['label'].map({
    'Household' : 0,
    'Books': 1,
    'Electronics': 2,
    'Clothing & Accessories': 3
})

df.head()
```

```python
# Spliting the data
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
    df.Text,
    df.label_num,
    test_size=0.2,
    random_state=2022,
    stratify= df.label_num
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(y_train.shape)
print(y_train.value_counts())

print(y_test.shape)
print(y_test.value_counts())
```

### Now we train different models on our data to pick the best one (Without Preprocessing the text)

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
from sklearn.feature_extraction.text import TfidfVectorizer

# Firstly trying with KNN
clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('knn',KNeighborsClassifier())
])

clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test,y_pred))
```

```python
# Just checking the first 5 predictions and comparing it with actual predictions
print(y_pred[:5])
print(y_test[:5])
```

```python
# Now tring MultinomialNB Classifier
from sklearn.naive_bayes import MultinomialNB

clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('nb',MultinomialNB())
])

clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test,y_pred))
```

```python
# Now RandomForest Classifier
from sklearn.ensemble import RandomForestClassifier

clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('randomForest',RandomForestClassifier())
])

clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test,y_pred)) # Giving Slightly Good performance from others (till now)
```

### Now By doing preprocessing we Again check the results for all models

```python

import spacy

nlp = spacy.load("en_core_web_sm")

def preprocess(text): # It will remove the stop words and punctuations and also Lemmatize the text.
    doc = nlp(text)
    preprocess_text=[ token.lemma_ for token in doc if not token.is_stop and not token.is_punct ]
    return " ".join(preprocess_text)
```

```python
df['preprocessed_text'] = df['Text'].apply(preprocess)
df.head()
```

```python
# Again spliting the data but this time with the Preprocessed_Text
X_train,X_test,y_train,y_test = train_test_split(
    df.preprocessed_text,
    df.label_num,
    test_size=0.2,
    random_state= 2022,
    stratify= df.label_num
)
```

```python
# KNN
clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('knn',KNeighborsClassifier())
])

clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test,y_pred))
```

```python
# MultinomialNB
clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('nb',MultinomialNB())
])

clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test,y_pred))
```

```python
# Randomforest
clf = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('randomForest',RandomForestClassifier())
])

clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test,y_pred))
```
