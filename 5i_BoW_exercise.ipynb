# Exercise of Bag Of Words
* In this Exercise, you are going to classify whether a given movie review is positive or negative.
* You are going to use Bag of words for pre-processing the text and apply different classification algorithms.
* Sklearn CountVectorizer has the inbuilt implementations for Bag of Words.

```python
#1. read the data provided in the same directory with name 'movies_sentiment_data.csv' and store it in df variable
import pandas as pd

df = pd.read_csv('movies_sentiment_data.csv')

#2. print the shape of the data
print(f"Shape of the data frame : {df.shape}")

#3. print top 5 datapoints
df.head()
```

```python
#Creating a new column "Category" which represent 1 if the sentiment is positive or 0 if it is negative

# Approach 1 using Lambda Function (Just another way to write like 2nd approach)
# df['Category'] = df['sentiment'].apply(lambda x: 1 if x=='positive' else 0)

# Approach 2 using map (More Controlled and Preferred) - Here we can add as many category as we want.
df['Category'] = df['sentiment'].map({
    'negative':0,
    'positive':1
})

# Approach 2 using LabelEncoder
# from sklearn.preprocessing import LabelEncoder

# encoder = LabelEncoder()
# Category = encoder.fit_transform(df.sentiment)
# df['Category'] = Category


df.head()
```

```python
#check the distribution of 'Category' and see whether the Target labels are balanced or not.
df['Category'].value_counts()
```

```python
#Do the 'train-test' splitting with test size of 20%
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(df['review'],df['Category'],test_size=0.2)

print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

Exercise-1

using sklearn pipeline module create a classification pipeline to classify the movie review's positive or negative.

Note:

use CountVectorizer for pre-processing the text.

use Random Forest as the classifier with estimators as 50 and criterion as entropy.

print the classification report.


References:

https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html

https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

classifier = Pipeline([
    ('vectorizer',CountVectorizer()),
    ('RandomForestClassifier',RandomForestClassifier(n_estimators=50,criterion='entropy'))
])

classifier.fit(X_train,y_train)

y_pred_clf = classifier.predict(X_test)
print(classification_report(y_test,y_pred_clf))
```

* Our RandomForestClassifier Model is giving us 84% Accuracy.

# How to read this report and what each component is telling
* Precision: Out of everything |the model predicted as positive, how much was actually positive?
* Recall: Out of everything that was actually positive, how much did the model successfully find?
* F1-score: Combination of Precision + Recall (using their harmonic mean). If both precision and recall are good, F1 will also be good.

F1 = 2 × (Precision × Recall)  /  (Precision + Recall)

* Support → Number of actual examples - How many actual samples belong to each class

* using sklearn pipeline module create a classification pipeline to classify the movie review's positive or negative..

Note:

* use CountVectorizer for pre-processing the text.
* use Multinomial Naive Bayes as the classifier.
* print the classification report.

References:

https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html

https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.MultinomialNB.html

```python
from sklearn.naive_bayes import MultinomialNB
nb_clf = Pipeline([
    ('vectorizer',CountVectorizer()),
    ('nb',MultinomialNB())
])

nb_clf.fit(X_train,y_train)

y_pred_nb_clf= nb_clf.predict(X_test)
print(classification_report(y_test,y_pred_nb_clf))
```

# Can you write some observations of why model like KNN fails to produce good results unlike RandomForest and MultinomialNB ?

KNN = "Find similar examples."
* High-dimensional text (bcz of BoW) → calculating distances becomes difficult → may perform poorly.

Multinomial Naive Bayes = "Which words are likely for each class?"
* Works naturally with word counts/frequencies → often useful for text classification.

Random Forest = "Many decision trees vote together."
* Uses different samples/features → can handle many features and capture useful patterns.

Final point
* Don't assume an algorithm will work best. Try suitable algorithms, evaluate them using metrics like F1/accuracy/precision/recall and practical requirements, then choose based on the evidence.
