## <font color='yellow'> Word Vector Using Gensim Library </font>

```python
import gensim.downloader as api
```

```python
# All gensim models are listed on this page: https://github.com/RaRe-Technologies/gensim-data
wv = api.load('word2vec-google-news-300')
print("Model is ready to work!")
```

```python
print(wv.similarity(w1="great",w2="good")) # Similiarity function tell us how much similiar the both words acc. to their context they used with
print(wv.similarity(w1="bread",w2="butter"))
print(wv.similarity(w1="good",w2="bad"))
```

```python
# We also have a function named most_similiar() that tells the most similiar words to a provided word
print(wv.most_similar("good"))
print(wv.most_similar("bread"))
print(wv.most_similar("dog",topn= 5))
```

```python
# We can also do the equation like examples
print(wv.most_similar(positive=['king','woman'],negative=['man'])) # King - Man + Woman = Queen
print(wv.most_similar(positive=['france','berlin'],negative=['paris'])) # France - Paris + Berlin = Germany
```

```python
# doesnt_match() function 
print(wv.doesnt_match(["facebook","google","cat","microsoft"])) 
print(wv.doesnt_match(["dog", "cat", "google", "mouse"])) 
```
