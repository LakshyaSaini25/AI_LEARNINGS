Customization of Spacy Object

```python
from spacy.symbols import ORTH

nlp.tokenizer.add_special_case("gimme",[
    {ORTH: "gim"},
    {ORTH: "me"}
])

doc = nlp("gimme double cheeze extra large pizza") # Now gim and me are two seperate tokens

for token in doc:
    print(token)
```

# Stemming using NLTK

# Customization in Language Processing Pipelining

```python
ar = nlp.get_pipe('attribute_ruler')
ar.add([[{"TEXT": "Bro"}] , [{"TEXT": "Bruh"}]],{"LEMMA":"Brother"})

doc = nlp("Bro, Your name is Bruh")
for token in doc:
    print(token.text, "|", token.lemma_)
```

```python
# Counting the Part of Speech in the Text using spacy's count_by API

import spacy.attrs


count = doc.count_by(spacy.attrs.POS)
print(count)

for k,v in count.items():
    print(doc.vocab[k].text,"|",v)
```

# Exercise 
Exercise for Spacy POS tutorial,

You are parsing a news story from cnbc.com. News story is stores in news_story.txt which is available in this same folder on github. You need to,
* Extract all NOUN tokens from this story. You will have to read the file in python first to collect all the text and then extract NOUNs in a python list
* Extract all numbers (NUM POS type) in a python list
* Print a count of all POS tags in this story

```python
with open("news_story.txt","r") as f:
    news_text = f.read()

news_text[:500]
```

```python
# 1 and 2) Extract all NOUN tokens and all the NUM tokens
doc = nlp(news_text)

noun_tokens=[]
num_tokens =[]
for token in doc:
    if(token.pos_ == "NOUN"):
        noun_tokens.append([token.text,token.pos_])
    if(token.pos_ == "NUM"):
        num_tokens.append([token.text,token.pos_])

print("Noun Tokens:",noun_tokens[:10])
print("Number Tokens:",num_tokens[:10])
print("Count of all the POS:")

# Count of all the POS in the story
count = doc.count_by(spacy.attrs.POS)
for k,v in count.items():
    print(doc.vocab[k].text,"|",v)
```
