# Bag Of Words (BoW)  
* Algo to convert text into Numerical representation
* Bag of words (a.k.a. BOW) is a technique used for text representation in natural language processing. In this NLP tutorial, we will go over how a bag of words works and also write some code for email classification that uses a bag of words and the Naive Bayes classifier in machine learning.

```python
# %pip install pandas
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('spam.csv')
df.head()
```

```python
print(df.Category.value_counts()) 
# So, Firstly let's convert the Category column (As it has only 2 distinct value) into a simple Numerical representation of 0 and 1


# Instead of using lambda funtion we can also use below function
def get_spam_number(x):  # and call this particular function from apply(get_spam_number)
    if x=='spam':
        return 1
    return 0

# Making a new column named (spam) 
df['spam'] = df['Category'].apply(lambda x: 1 if x=='spam' else 0)
# df['spam'] = df['Category'].apply(get_spam_number)

print(f"Shape of the given dataframe : {df.shape}")
df.head()
```

```python
# Now spliting the data into Test-Train
# %pip install scikit-learn
from sklearn.model_selection import train_test_split

# The reason of 
X_train,X_test,y_train,y_test = train_test_split(df.Message,df.spam,test_size=0.2)
```

```python
# Converting the X_train from Text Representation to Vector so that we can feed that into our ML model
from sklearn.feature_extraction.text import CountVectorizer # We are using Bag Of Words Algo

v = CountVectorizer()

X_train_cv = v.fit_transform(X_train.values) # X_train_cv => cv stands for countVectorized
print(X_train_cv.toarray())
X_train_cv.shape # It means we have 7753 unique words in our vocabulary
```

```python
v.vocabulary_ # By this we can see all the unique words in our made vocabulary
```

```python
# Now let us train our ML model with the training data 
# We are using Naive Bayes Classifier also in that we are using MultinomialNB where NB stands for Naive Bayes

from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_cv,y_train) # Now our model is trained so we test that.
```

```python
# First let's also convert X_test into X_test_cv
X_test_cv = v.transform(X_test)
X_test_cv.toarray()
```

```python
# Now testing our model and Compare the results using classification_report

from sklearn.metrics import classification_report

y_pred = model.predict(X_test_cv)

print(classification_report(y_test,y_pred)) # First providing the Actual truth i.e y_test and then Our predicted result y_pred
```

# Differ bw  fit , transform and fit_transform

```python
# fit() -> LEARN = makes the object learn the required information/pattern from the data
# Example when we using in Bag of Word
v.fit(X_train) # CountVectorizer looks at X_train and learns the vocabulary.

# But fit() doesn't convert the sentences into numbers.

X_train_cv = v.transform(X_train) # transform()= CONVERT = takes data and converts it according to what was learned during fit().
# The previously learned vocabulary is used to convert text into numerical vectors.
print(X_train_cv.toarray())

# fit_transform() = LEARN + CONVERT =  It performs both operations together:
v.fit_transform(X_train) # Instead of seperately doing do it at once
```

# We use fit_transform() for Training Data and transform() on Test Data ? Why?

We want the vectorizer to:

* Learn vocabulary from training data.
* Convert training data into numerical form.

so, X_train -> fit -> Learn vocabulary -> transform -> Numerical data

And for the test data we are using X_test_cv = v.transform(X_test) bcz
* We don't want to learn anything from the test data.
* The test data should be treated as unseen data.

```python
# Complete Example of Bag Of Word
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

v = CountVectorizer()

# Training
X_train_cv = v.fit_transform(X_train)

# Train model
model = MultinomialNB()
model.fit(X_train_cv, y_train)

# Testing
X_test_cv = v.transform(X_test)

# Prediction
y_pred = model.predict(X_test_cv)
```

# Now Instead of doing all of this manually we can use sklearn.pipeline to Done whole procedure

```python
from sklearn.pipeline import Pipeline

clf = Pipeline([ # clf is the name of the variable stands for classifer and we can give any name.
    ('vectorizer',CountVectorizer()), # The first one is the name of the step (We can give any name here instead of vectorizer) and the second one is the Object of what we need to give to the pipeline
    ('nb',MultinomialNB())
])

clf.fit(X_train,y_train) # This time we don't need to provide the X_train_cv as The CountVectorizer in the pipeline will handle the same and convert it into vectors automatically and feed that into our ML model in the next step of the pipeline

y_pred_clf = clf.predict(X_test)
print(classification_report(y_test,y_pred_clf))
```
